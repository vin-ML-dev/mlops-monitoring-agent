# Explore the curated data
Run `make data` first to generate the files.

```python
import collections
import json
import pathlib

root = pathlib.Path('..')
report = json.load(open(root / 'data/interim/curation_report.json'))
print('kept/dropped counts:', report['counts'])
print('PII removed:', report['pii_removed'])
```

```python
# peek at a few training records
rows = [json.loads(line) for line in open(root / 'data/processed/train.jsonl')]
print('train size:', len(rows))
for r in rows[:2]:
    print(r['meta'])
    for m in r['messages']:
        print(' ', m['role'], '::', m['content'][:80])
```

```python
# kind mix across the training split (finance vs honesty examples)
kinds = collections.Counter(r['meta'].get('kind') for r in rows)
print(kinds)
```
